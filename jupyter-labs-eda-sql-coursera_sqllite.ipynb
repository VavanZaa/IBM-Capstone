### Connect to the SQLite database

Run all cells in order. This SQLite edition executes ordinary SQL using `sqlite3` and displays query results with pandas. It does not require SQLAlchemy, `ipython-sql`, or a PrettyTable workaround.

The unchanged course SQL dataset is supplied at `data/Spacex.csv` (101 rows). It differs from the 90-row API/wrangling dataset. All queries below run locally; no internet connection is required once pandas is installed.

```python
import sqlite3
from pathlib import Path
import pandas as pd
```

```python
# Support running from the project root or from this notebook's folder.
LAB_DIR = Path("capstone") if Path("capstone").is_dir() else Path(".")
dataset_path = LAB_DIR / "data" / "Spacex.csv"
database_path = LAB_DIR / "my_data1.db"
```

```python
# Close this notebook's previous connection if the setup cell is rerun.
if isinstance(globals().get("con"), sqlite3.Connection):
    con.close()
con = sqlite3.connect(database_path)
cur = con.cursor()
print(f"Connected to SQLite: {database_path}")
```

```python
def run_sql(query, params=None):
    """Execute a SQL SELECT against the same SQLite connection used for loading."""
    return pd.read_sql_query(query, con, params=params)
```

```python
run_sql("SELECT sqlite_version() AS SQLite_Version")
```

```python
if not dataset_path.is_file():
    raise FileNotFoundError(
        f"Missing {dataset_path}. Keep the supplied data folder beside the notebook."
    )
df = pd.read_csv(dataset_path)
required_columns = {
    "Date", "Time (UTC)", "Booster_Version", "Launch_Site", "Payload",
    "PAYLOAD_MASS__KG_", "Orbit", "Customer", "Mission_Outcome", "Landing_Outcome",
}
if required_columns - set(df.columns):
    raise ValueError(f"Missing SQL lab columns: {sorted(required_columns - set(df.columns))}")
if df.empty:
    raise ValueError("The SQL lab dataset is empty.")
# Preserve the source values in the raw table.
df.to_sql("SPACEXTBL", con, if_exists="replace", index=False)
print(f"Loaded {len(df)} records from {dataset_path}")
df.head()
```

Create the analysis table from the raw table, excluding null or blank dates and trimming leading/trailing whitespace from text fields. The original CSV and raw table retain their source values.

```python
# Recreate only this lab's derived table; rerunning does not duplicate rows.
with con:
    con.execute("DROP TABLE IF EXISTS SPACEXTABLE")
```

```python
with con:
    con.execute('''
        CREATE TABLE SPACEXTABLE AS
        SELECT TRIM("Date") AS "Date",
               TRIM("Time (UTC)") AS "Time (UTC)",
               TRIM(Booster_Version) AS Booster_Version,
               TRIM(Launch_Site) AS Launch_Site,
               TRIM(Payload) AS Payload,
               PAYLOAD_MASS__KG_,
               TRIM(Orbit) AS Orbit,
               TRIM(Customer) AS Customer,
               TRIM(Mission_Outcome) AS Mission_Outcome,
               TRIM(Landing_Outcome) AS Landing_Outcome
        FROM SPACEXTBL
        WHERE "Date" IS NOT NULL AND TRIM("Date") <> ''
    ''')
run_sql("SELECT COUNT(*) AS Launch_Count FROM SPACEXTABLE")
```

## Tasks

Now write and execute SQL queries to solve the assignment tasks.

**Note: If the column names are in mixed case enclose it in double quotes
   For Example "Landing_Outcome"**

### Task 1




##### Display the names of the unique launch sites  in the space mission


```python
task1 = run_sql('''
    SELECT DISTINCT Launch_Site
    FROM SPACEXTABLE
    ORDER BY Launch_Site
''')
task1
```


### Task 2


#####  Display 5 records where launch sites begin with the string 'CCA' 


```python
task2 = run_sql('''
    SELECT *
    FROM SPACEXTABLE
    WHERE Launch_Site LIKE 'CCA%'
    ORDER BY "Date", "Time (UTC)"
    LIMIT 5
''')
task2
```

### Task 3




##### Display the total payload mass carried by boosters launched by NASA (CRS)


```python
# Match the NASA (CRS) customer category exactly, as specified.
task3 = run_sql('''
    SELECT SUM(PAYLOAD_MASS__KG_) AS Total_Payload_Mass_KG
    FROM SPACEXTABLE
    WHERE Customer = 'NASA (CRS)'
''')
task3
```

### Task 4




##### Display average payload mass carried by booster version F9 v1.1


```python
# Exact dataset label: serial-number variants are separate Booster_Version values.
task4 = run_sql('''
    SELECT AVG(PAYLOAD_MASS__KG_) AS Average_Payload_Mass_KG
    FROM SPACEXTABLE
    WHERE Booster_Version = 'F9 v1.1'
''')
task4
```

### Task 5

##### List the date when the first succesful landing outcome in ground pad was acheived.


_Hint:Use min function_ 


```python
task5 = run_sql('''
    SELECT MIN("Date") AS First_Successful_Ground_Pad_Landing
    FROM SPACEXTABLE
    WHERE Landing_Outcome = 'Success (ground pad)'
''')
task5
```

### Task 6

##### List the names of the boosters which have success in drone ship and have payload mass greater than 4000 but less than 6000


```python
# Strict bounds: payloads equal to 4000 or 6000 kg are excluded.
task6 = run_sql('''
    SELECT DISTINCT Booster_Version
    FROM SPACEXTABLE
    WHERE Landing_Outcome = 'Success (drone ship)'
      AND PAYLOAD_MASS__KG_ > 4000
      AND PAYLOAD_MASS__KG_ < 6000
    ORDER BY Booster_Version
''')
task6
```

### Task 7




##### List the total number of successful and failure mission outcomes


```python
# Count mission outcomes, not booster landing outcomes.
# "Success (payload status unclear)" counts as Success according to its source label.
task7 = run_sql('''
    SELECT CASE
               WHEN Mission_Outcome LIKE 'Success%' THEN 'Success'
               WHEN Mission_Outcome LIKE 'Failure%' THEN 'Failure'
               ELSE 'Other / unknown'
           END AS Mission_Result,
           COUNT(*) AS Mission_Count
    FROM SPACEXTABLE
    GROUP BY Mission_Result
    ORDER BY Mission_Count DESC, Mission_Result
''')
task7
```

### Task 8



##### List all the booster_versions that have carried the maximum payload mass, using a subquery with a suitable aggregate function.


```python
task8 = run_sql('''
    SELECT DISTINCT Booster_Version, PAYLOAD_MASS__KG_
    FROM SPACEXTABLE
    WHERE PAYLOAD_MASS__KG_ = (
        SELECT MAX(PAYLOAD_MASS__KG_) FROM SPACEXTABLE
    )
    ORDER BY Booster_Version
''')
task8
```

### Task 9


##### List the records which will display the month names, failure landing_outcomes in drone ship ,booster versions, launch_site for the months in year 2015.

**Note: SQLLite does not support monthnames. So you need to use  substr(Date, 6,2) as month to get the months and substr(Date,0,5)='2015' for year.**


```python
# SQLite substr extracts the month; CASE supplies the requested month name.
task9 = run_sql('''
    SELECT CASE SUBSTR("Date", 6, 2)
               WHEN '01' THEN 'January'
               WHEN '02' THEN 'February'
               WHEN '03' THEN 'March'
               WHEN '04' THEN 'April'
               WHEN '05' THEN 'May'
               WHEN '06' THEN 'June'
               WHEN '07' THEN 'July'
               WHEN '08' THEN 'August'
               WHEN '09' THEN 'September'
               WHEN '10' THEN 'October'
               WHEN '11' THEN 'November'
               WHEN '12' THEN 'December'
           END AS Month_Name,
           Landing_Outcome, Booster_Version, Launch_Site
    FROM SPACEXTABLE
    WHERE SUBSTR("Date", 1, 4) = '2015'
      AND Landing_Outcome = 'Failure (drone ship)'
    ORDER BY "Date", "Time (UTC)"
''')
task9
```

### Task 10




##### Rank the count of landing outcomes (such as Failure (drone ship) or Success (ground pad)) between the date 2010-06-04 and 2017-03-20, in descending order.


```python
# ISO dates sort chronologically; BETWEEN includes both requested endpoints.
task10 = run_sql('''
    SELECT Landing_Outcome, COUNT(*) AS Outcome_Count
    FROM SPACEXTABLE
    WHERE "Date" BETWEEN '2010-06-04' AND '2017-03-20'
    GROUP BY Landing_Outcome
    ORDER BY Outcome_Count DESC, Landing_Outcome
''')
task10
```
