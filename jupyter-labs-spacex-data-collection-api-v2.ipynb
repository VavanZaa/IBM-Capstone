## Import Libraries and Define Auxiliary Functions


We will import the following libraries into the lab


```python
import requests
import pandas as pd
import numpy as np
import datetime
import json
from functools import lru_cache
from pathlib import Path

pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', None)
DATA_DIR = Path("capstone/data") if Path("capstone").is_dir() else Path("data")
# Historical lab mode avoids waiting for the unavailable live API.
# Set True and restart the kernel to attempt live API requests.
USE_LIVE_API = False
session = requests.Session()
_reported_sources = set()


@lru_cache(maxsize=1)
def load_metadata_snapshot():
    path = DATA_DIR / "spacex_lab_metadata.json"
    if not path.is_file():
        raise FileNotFoundError(
            f"Missing {path}. Keep this notebook beside its supplied data folder."
        )
    return json.loads(path.read_text())


def get_snapshot_record(resource, record_id):
    snapshot = load_metadata_snapshot()
    record = snapshot["records"].get(resource, {}).get(record_id)
    if record is not None:
        if resource not in _reported_sources:
            print(f"{resource}: using documented historical lab metadata.")
            _reported_sources.add(resource)
        return record
    excluded = snapshot["excluded_ids"].get(record_id)
    if excluded and excluded["resource"] == resource:
        # These four Falcon 1 launches are removed in Task 2.
        # Empty metadata keeps the rows aligned without invented values.
        if "excluded" not in _reported_sources:
            print("Unavailable Falcon 1 core/launchpad details remain missing; "
                  "Task 2 removes those launches.")
            _reported_sources.add("excluded")
        return {}
    raise KeyError(f"No verified {resource} metadata for ID {record_id!r}.")


@lru_cache(maxsize=None)
def get_api_record(resource, record_id):
    """Resolve unique IDs using historical data, or live HTTP with a fallback."""
    if not USE_LIVE_API:
        return get_snapshot_record(resource, record_id)
    url = f"https://api.spacexdata.com/v4/{resource}/{record_id}"
    try:
        response = session.get(url, timeout=(5, 10))
        response.raise_for_status()
        return response.json()
    except (requests.RequestException, ValueError) as exc:
        try:
            return get_snapshot_record(resource, record_id)
        except (FileNotFoundError, KeyError) as fallback_error:
            raise RuntimeError(
                f"Cannot retrieve {resource}/{record_id}: the live request failed "
                f"({exc}) and no local record is available ({fallback_error})."
            ) from exc
```

Below we will define a series of helper functions that will help us use the API to extract information using identification numbers in the launch data.

From the <code>rocket</code> column we would like to learn the booster name.


```python
def getBoosterVersion(data):
    if "rocket" not in data:
        raise ValueError(
            "This cell requires the launch dataframe with its rocket ID column. "
            "Restart the kernel and run all cells in order."
        )
    # Replace previous results when rerun; never append duplicate launches.
    BoosterVersion[:] = [
        get_api_record("rockets", rocket)["name"] if rocket else None
        for rocket in data["rocket"]
    ]
```

From the <code>launchpad</code> we would like to know the name of the launch site being used, the logitude, and the latitude.


```python
def getLaunchSite(data):
    records = [
        get_api_record("launchpads", pad) if pad else {}
        for pad in data["launchpad"]
    ]
    Longitude[:] = [record.get("longitude") for record in records]
    Latitude[:] = [record.get("latitude") for record in records]
    LaunchSite[:] = [record.get("name") for record in records]
```

From the <code>payload</code> we would like to learn the mass of the payload and the orbit that it is going to.


```python
def getPayloadData(data):
    records = [
        get_api_record("payloads", payload) if payload else {}
        for payload in data["payloads"]
    ]
    PayloadMass[:] = [record.get("mass_kg") for record in records]
    Orbit[:] = [record.get("orbit") for record in records]
```

From <code>cores</code> we would like to learn the outcome of the landing, the type of the landing, number of flights with that core, whether gridfins were used, wheter the core is reused, wheter legs were used, the landing pad used, the block of the core which is a number used to seperate version of cores, the number of times this specific core has been reused, and the serial of the core.


```python
def getCoreData(data):
    cores = list(data["cores"])
    records = [
        get_api_record("cores", core["core"]) if core.get("core") else {}
        for core in cores
    ]
    Block[:] = [record.get("block") for record in records]
    ReusedCount[:] = [record.get("reuse_count") for record in records]
    Serial[:] = [record.get("serial") for record in records]
    Outcome[:] = [
        f"{core.get('landing_success')} {core.get('landing_type')}"
        for core in cores
    ]
    Flights[:] = [core.get("flight") for core in cores]
    GridFins[:] = [core.get("gridfins") for core in cores]
    Reused[:] = [core.get("reused") for core in cores]
    Legs[:] = [core.get("legs") for core in cores]
    LandingPad[:] = [core.get("landpad") for core in cores]
```

### Task 1: Request and parse the SpaceX launch data using the GET request


To make the requested JSON results more consistent, we will use the following static response object for this project:


```python
static_json_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DS0321EN-SkillsNetwork/datasets/API_call_spacex_api.json'
snapshot = DATA_DIR / "API_call_spacex_api.json"
response = None
if not USE_LIVE_API and snapshot.is_file():
    launch_json = json.loads(snapshot.read_text())
    print(f"Loaded unchanged course launch snapshot: {snapshot}")
else:
    try:
        response = session.get(static_json_url, timeout=30)
        response.raise_for_status()
        launch_json = response.json()
        print(f"Static launch snapshot: HTTP {response.status_code}")
    except requests.RequestException:
        response = None
        if not snapshot.is_file():
            raise
        launch_json = json.loads(snapshot.read_text())
        print(f"Network unavailable; using the unchanged course snapshot: {snapshot}")
```

We will now use the API again to get information about the launches using the IDs given for each launch. Specifically we will be using columns <code>rocket</code>, <code>payloads</code>, <code>launchpad</code>, and <code>cores</code>.


```python
data = data[['rocket', 'payloads', 'launchpad', 'cores', 'flight_number', 'date_utc']].copy()
# Keep launches with exactly one core and one payload.
data = data.loc[
    (data['cores'].map(len) == 1) & (data['payloads'].map(len) == 1)
].copy()
data['cores'] = data['cores'].map(lambda values: values[0])
data['payloads'] = data['payloads'].map(lambda values: values[0])
data['date'] = pd.to_datetime(data['date_utc'], utc=True).dt.date
data = data.loc[data['date'] <= datetime.date(2020, 11, 13)].copy()
```

```python
# Call getBoosterVersion.
# Refresh the lookup here as well, so this cell repairs an older running kernel.
# Live mode remains available when explicitly selected in the setup cell.
if not globals().get("USE_LIVE_API", False):
    import json
    from pathlib import Path
    from functools import lru_cache

    metadata_path = Path("data/spacex_lab_metadata.json")
    if not metadata_path.is_file():
        metadata_path = Path("capstone/data/spacex_lab_metadata.json")
    if not metadata_path.is_file():
        raise FileNotFoundError(
            "Keep the supplied data folder beside this notebook, then rerun this cell."
        )
    metadata = json.loads(metadata_path.read_text())

    @lru_cache(maxsize=None)
    def get_api_record(resource, record_id):
        records = metadata["records"].get(resource, {})
        if record_id in records:
            return records[record_id]
        excluded = metadata["excluded_ids"].get(record_id, {})
        if excluded.get("resource") == resource:
            return {}  # Unavailable Falcon 1 details; removed by Task 2.
        raise KeyError(f"No verified {resource} metadata for ID {record_id!r}.")

    print("Using documented historical metadata; no API request is needed.")

getBoosterVersion(data)
print(f"Loaded booster names for {len(BoosterVersion)} launches.")
print(pd.Series(BoosterVersion).value_counts())
```

### Task 3: Dealing with Missing Values


Calculate below the mean for the <code>PayloadMass</code> using the <code>.mean()</code>. Then use the mean and the <code>.replace()</code> function to replace `np.nan` values in the data with the mean you calculated.


```python
# Calculate the mean from observed Falcon 9 payload masses only.
payload_mean = data_falcon9['PayloadMass'].mean()
if pd.isna(payload_mean):
    raise ValueError("No observed payload masses are available to calculate the mean.")
data_falcon9['PayloadMass'] = data_falcon9['PayloadMass'].replace(np.nan, payload_mean)
assert data_falcon9['PayloadMass'].notna().all()
data_falcon9.isnull().sum()
```
